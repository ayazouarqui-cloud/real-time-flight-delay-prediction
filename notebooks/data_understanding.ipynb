```python
import pandas as pd
import numpy as np

pd.set_option('display.max_rows', 100)
pd.set_option('display.max_columns', 100)
pd.set_option('display.width', 150)


df = pd.read_csv('../Combined_Flights_2019.csv')
```

```python
manquants = df.isna().sum()
manquants = manquants[manquants > 0].sort_values(ascending=False)
pct = (manquants / len(df) * 100).round(2)
pd.DataFrame({'n_manquant': manquants, '%': pct})

n_cancel = df['Cancelled'].sum()
n_divert = df['Diverted'].sum()
print(f"Annulés  : {n_cancel:,} ({n_cancel/len(df):.2%})")
print(f"Déroutés : {n_divert:,} ({n_divert/len(df):.2%})")

print("\nManquant DepTime vs Cancelled:", df['DepTime'].isna().sum(), "vs", n_cancel)
print("Ont WheelsOn mais pas ArrDelay (≈ déroutés):",
      df[df['WheelsOn'].notna() & df['ArrDelay'].isna()].shape[0], "vs", n_divert)

vols_normaux = df[~df['Cancelled'] & ~df['Diverted']]
reste = vols_normaux.isna().sum()
reste[reste > 0].sort_values(ascending=False)
```

```python
incoherences = {}

incoherences['CRSElapsedTime < 0'] = (df['CRSElapsedTime'] < 0).sum()
incoherences['ActualElapsedTime <= 0'] = (df['ActualElapsedTime'] <= 0).sum()
incoherences['AirTime <= 0'] = (df['AirTime'] <= 0).sum()
incoherences['Distance <= 0'] = (df['Distance'] <= 0).sum()
incoherences['TaxiOut < 0'] = (df['TaxiOut'] < 0).sum()
incoherences['TaxiIn < 0'] = (df['TaxiIn'] < 0).sum()
incoherences['CRSDepTime hors [1,2400]'] = (~df['CRSDepTime'].between(1, 2400)).sum()
incoherences['CRSArrTime hors [1,2400]'] = (~df['CRSArrTime'].between(1, 2400)).sum()

pd.Series(incoherences)

df[df['CRSElapsedTime'] < 0][
    ['FlightDate', 'Origin', 'Dest', 'CRSDepTime', 'CRSArrTime', 'CRSElapsedTime']
].head(10)
```

```python
def verifier_redondance(df, col_a, col_b):
    n_a = df[col_a].nunique()
    n_b = df[col_b].nunique()
    n_paires = df[[col_a, col_b]].drop_duplicates().shape[0]
    est_1_1 = (n_paires == n_a) and (n_paires == n_b)
    print(f"{col_a} ({n_a}) vs {col_b} ({n_b}) -> {n_paires} paires uniques "
          f"-> {'REDONDANT (1:1)' if est_1_1 else 'PAS 1:1, vérifier'}")

paires_a_verifier = [
    ('Origin', 'OriginAirportID'),
    ('Origin', 'OriginAirportSeqID'),
    ('Origin', 'OriginCityName'),
    ('Origin', 'OriginCityMarketID'),
    ('OriginState', 'OriginStateFips'),
    ('OriginState', 'OriginStateName'),
    ('Dest', 'DestAirportID'),
    ('Dest', 'DestAirportSeqID'),
    ('Dest', 'DestCityName'),
    ('DestState', 'DestStateFips'),
    ('DestState', 'DestStateName'),
    ('Airline', 'IATA_Code_Marketing_Airline'),
    ('Airline', 'Marketing_Airline_Network'),
    ('IATA_Code_Operating_Airline', 'Operating_Airline'),
    ('DepDelay', 'DepDelayMinutes'),
]
for a, b in paires_a_verifier:
    verifier_redondance(df, a, b)

connu_avant_depart = [
    'FlightDate', 'Airline', 'Origin', 'Dest', 'CRSDepTime', 'CRSArrTime',
    'CRSElapsedTime', 'Distance', 'DistanceGroup', 'Year', 'Quarter', 'Month',
    'DayofMonth', 'DayOfWeek', 'DepTimeBlk', 'ArrTimeBlk',
]

connu_apres_decollage_seulement = [
    'DepTime', 'DepDelay', 'DepDelayMinutes', 'DepDel15',
    'DepartureDelayGroups', 'TaxiOut', 'WheelsOff',
]

jamais_utilisable_pour_predire_le_retard = [
    'WheelsOn', 'TaxiIn', 'ArrTime', 'ArrDelay', 'ArrDelayMinutes',
    'ArrDel15', 'ArrivalDelayGroups', 'AirTime', 'ActualElapsedTime',
]

print("AVANT décollage :", connu_avant_depart)
print("\nEN PLUS si prédiction juste APRÈS décollage :", connu_apres_decollage_seulement)
print("\nJAMAIS utilisable comme feature (fuite de données) :", jamais_utilisable_pour_predire_le_retard)
```
